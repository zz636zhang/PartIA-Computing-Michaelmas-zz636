### 1. Dot Product

For two vectors of length $n$, the dot product requires $n$ multiplications and $n-1$ additions.

The total number of operations is:

$$
n+(n-1)=2n-1
$$

Therefore, the computational complexity is:

$$
O(n)
$$

### 2. Matrix-Vector Product

For an $n \times n$ matrix and a vector of length $n$, each row requires one dot product.

There are $n$ rows, so the total number of operations is:

$$
n(2n-1)=2n^2-n
$$

Therefore, the computational complexity is:

$$
O(n^2)
$$

### 3. Matrix-Matrix Product

For two $n \times n$ matrices, the resulting matrix contains $n^2$ elements.

Each element requires one dot product of length $n$.

The total number of operations is:

$$
n^2(2n-1)=2n^3-n^2
$$

Therefore, the computational complexity of standard matrix multiplication is:

$$
O(n^3)
$$

### Conclusion

The computational complexity increases from $O(n)$ for the dot product, to $O(n^2)$ for the matrix-vector product, and $O(n^3)$ for standard matrix-matrix multiplication.

## Exercise 11.2

For the recursive factorial algorithm in Activity 04, determine the algorithmic complexity by inspecting your implementation of the algorithm. Test this against numerical experiments.

### Solution

Recall the factorial algorithm from Activity 04.4:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return factorial(n - 1)*n
```

The function calls itself (recursively) $n$ times, hence it has complexity $O(n)$. We test this below and plot
the times.

```python
import time

n_values = np.arange(10, 501, 10)

times = []
for n in n_values:
    start = time.perf_counter()

    for i in range(1000):
        factorial(int(n))

    end = time.perf_counter()

    times.append((end - start) / 1000)

# Plot execution times
plt.plot(n_values, times)

plt.xlabel('n')
plt.ylabel('Average execution time (seconds)')
plt.title('Complexity of Recursive Factorial')
plt.grid(True)
plt.show()

```

### Solution

Time computation of determinant:

```python
# Create array of problem sizes we want to test (powers of 2)
N = 2**np.arange(2, 12)

times = []

for n in N:
    # Generate an n x n random matrix
    A = np.random.rand(n, n)

    start = time.perf_counter()
    det = np.linalg.slogdet(A)
    end = time.perf_counter()
    times.append(end - start)
```

Plot result:

```python
plt.plot(N, times, label='Measured time')

plt.xlabel('Matrix size (n)')
plt.ylabel('Execution time (seconds)')
plt.title('Complexity of Matrix Determinant')
plt.legend()
plt.grid(True)
plt.show()
```
